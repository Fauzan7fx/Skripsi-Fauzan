```python
from pathlib import Path
import pandas as pd

BASE_PATH = Path.cwd()
OUTPUT_PATH = BASE_PATH / "output"

FILE_HASIL = OUTPUT_PATH / "hasil_model_fixed_2025.xlsx"

print(FILE_HASIL)
```

```python
df = pd.read_excel(FILE_HASIL)

print(df.shape)
display(df.head())
```

```python
models_utama = [
    "Total_XGB",
    "XGB_Components",
    "MultiOutputTree_XGB",
    "RF_Components",
    "Multivariate_RF"
]

df_rmse = df[
    df["model"].isin(models_utama)
].copy()

display(df_rmse.head())
```

```python
tabel_rmse = (
    df_rmse
    .pivot_table(
        index=["province", "district"],
        columns="model",
        values="RMSE"
    )
    .reset_index()
)

# Urutkan kolom sesuai kebutuhan
tabel_rmse = tabel_rmse[
    [
        "province",
        "district",
        "Total_XGB",
        "XGB_Components",
        "MultiOutputTree_XGB",
        "RF_Components",
        "Multivariate_RF"
    ]
]

# Nama kolom lebih enak untuk tabel skripsi
tabel_rmse = tabel_rmse.rename(columns={
    "province": "Provinsi",
    "district": "Kabupaten/Kota",
    "Total_XGB": "Total XGB",
    "XGB_Components": "XGB Components",
    "MultiOutputTree_XGB": "Multi-output Tree XGB",
    "RF_Components": "RF Components",
    "Multivariate_RF": "Multivariate RF"
})

display(
    tabel_rmse.round(4)
)
```

```python
provinsi_label = {
    "jogja": "D.I. Yogyakarta",
    "jateng": "Jawa Tengah"
}

tabel_rmse["Provinsi"] = (
    tabel_rmse["Provinsi"]
    .map(provinsi_label)
)

display(
    tabel_rmse.round(4)
)
```

```python
kolom_model = [
    "Total XGB",
    "XGB Components",
    "Multi-output Tree XGB",
    "RF Components",
    "Multivariate RF"
]

tabel_rmse["Model RMSE Terendah"] = (
    tabel_rmse[kolom_model]
    .idxmin(axis=1)
)

tabel_rmse["RMSE Terendah"] = (
    tabel_rmse[kolom_model]
    .min(axis=1)
)

display(
    tabel_rmse.round(4)
)
```

```python
ringkasan_rmse = (
    tabel_rmse
    .groupby("Provinsi")[kolom_model]
    .mean()
    .reset_index()
)

display(
    ringkasan_rmse.round(4)
)
```

```python
file_output = OUTPUT_PATH / "tabel_RMSE_5_model_2025.xlsx"

with pd.ExcelWriter(
    file_output,
    engine="openpyxl"
) as writer:

    tabel_rmse.round(4).to_excel(
        writer,
        sheet_name="RMSE_per_kabupaten",
        index=False
    )

    ringkasan_rmse.round(4).to_excel(
        writer,
        sheet_name="Rata_rata_provinsi",
        index=False
    )

print("Tersimpan:", file_output)
```

```python
models_utama = [
    "Total_XGB",
    "XGB_Components",
    "MultiOutputTree_XGB",
    "RF_Components",
    "Multivariate_RF"
]

df_tau = df[
    df["model"].isin(models_utama)
].copy()

tabel_tau = (
    df_tau
    .pivot_table(
        index=["province", "district"],
        columns="model",
        values="taudcl"
    )
    .reset_index()
)

tabel_tau = tabel_tau[
    [
        "province",
        "district",
        "Total_XGB",
        "XGB_Components",
        "MultiOutputTree_XGB",
        "RF_Components",
        "Multivariate_RF"
    ]
]

tabel_tau = tabel_tau.rename(columns={
    "province": "Provinsi",
    "district": "Kabupaten/Kota",
    "Total_XGB": "Total XGB",
    "XGB_Components": "XGB Components",
    "MultiOutputTree_XGB": "Multi-output Tree XGB",
    "RF_Components": "RF Components",
    "Multivariate_RF": "Multivariate RF"
})

provinsi_label = {
    "jogja": "D.I. Yogyakarta",
    "jateng": "Jawa Tengah"
}

tabel_tau["Provinsi"] = (
    tabel_tau["Provinsi"]
    .map(provinsi_label)
)

display(tabel_tau.round(4))
```

```python
kolom_model = [
    "Total XGB",
    "XGB Components",
    "Multi-output Tree XGB",
    "RF Components",
    "Multivariate RF"
]

tabel_tau["Model Tau-b Tertinggi"] = (
    tabel_tau[kolom_model]
    .idxmax(axis=1)
)

tabel_tau["Tau-b Tertinggi"] = (
    tabel_tau[kolom_model]
    .max(axis=1)
)

display(tabel_tau.round(4))
```

```python
ringkasan_tau = (
    tabel_tau
    .groupby("Provinsi")[kolom_model]
    .mean()
    .reset_index()
)

display(ringkasan_tau.round(4))
```

```python
file_tau = OUTPUT_PATH / "tabel_Kendall_Tau_5_model_2025.xlsx"

with pd.ExcelWriter(
    file_tau,
    engine="openpyxl"
) as writer:

    tabel_tau.round(4).to_excel(
        writer,
        sheet_name="Tau_per_kabupaten",
        index=False
    )

    ringkasan_tau.round(4).to_excel(
        writer,
        sheet_name="Rata_rata_provinsi",
        index=False
    )


print("Tersimpan:", file_tau)
```

```python
df_ie40 = df[
    df["model"].isin(models_utama)
].copy()

tabel_ie40 = (
    df_ie40
    .pivot_table(
        index=["province", "district"],
        columns="model",
        values="IE40"
    )
    .reset_index()
)

tabel_ie40 = tabel_ie40[
    [
        "province",
        "district",
        "Total_XGB",
        "XGB_Components",
        "MultiOutputTree_XGB",
        "RF_Components",
        "Multivariate_RF"
    ]
]

tabel_ie40 = tabel_ie40.rename(columns={
    "province": "Provinsi",
    "district": "Kabupaten/Kota",
    "Total_XGB": "Total XGB",
    "XGB_Components": "XGB Components",
    "MultiOutputTree_XGB": "Multi-output Tree XGB",
    "RF_Components": "RF Components",
    "Multivariate_RF": "Multivariate RF"
})

tabel_ie40["Provinsi"] = (
    tabel_ie40["Provinsi"]
    .map(provinsi_label)
)

display(tabel_ie40.round(4))
```

```python
# ==========================================
# UBAH IE40 MENJADI PERSEN
# ==========================================

tabel_ie40_persen = tabel_ie40.copy()

# Ubah IE40 kelima model menjadi persen
for col in kolom_model:
    tabel_ie40_persen[col] = tabel_ie40_persen[col] * 100

display(tabel_ie40_persen.round(2))
```

```python
ringkasan_ie40 = (
    tabel_ie40
    .groupby("Provinsi")[kolom_model]
    .mean()
    .reset_index()
)

display(ringkasan_ie40.round(4))
```

```python
from openpyxl import load_workbook

file_evaluasi = OUTPUT_PATH / "tabel_evaluasi_model_2025.xlsx"

with pd.ExcelWriter(
    file_evaluasi,
    engine="openpyxl"
) as writer:

    # =========================
    # RMSE
    # =========================
    tabel_rmse.round(4).to_excel(
        writer,
        sheet_name="RMSE_per_kabupaten",
        index=False
    )

    ringkasan_rmse.round(4).to_excel(
        writer,
        sheet_name="RMSE_rata2_prov",
        index=False
    )

    # =========================
    # KENDALL TAU-B
    # =========================
    tabel_tau.round(4).to_excel(
        writer,
        sheet_name="Tau_per_kabupaten",
        index=False
    )

    ringkasan_tau.round(4).to_excel(
        writer,
        sheet_name="Tau_rata2_prov",
        index=False
    )

    # =========================
    # IE40
    # Simpan dalam bentuk proporsi 0-1
    # =========================
    tabel_ie40.to_excel(
        writer,
        sheet_name="IE40_per_kabupaten",
        index=False
    )

    ringkasan_ie40.to_excel(
        writer,
        sheet_name="IE40_rata2_prov",
        index=False
    )


# ==================================================
# FORMAT IE40 MENJADI PERSEN DI EXCEL
# ==================================================

wb = load_workbook(file_evaluasi)

# -------------------------
# IE40 per kabupaten
# -------------------------
ws = wb["IE40_per_kabupaten"]

# Kolom C-G = 5 model
for row in ws.iter_rows(
    min_row=2,
    min_col=3,
    max_col=7
):
    for cell in row:
        cell.number_format = "0.00%"

# -------------------------
# IE40 rata-rata provinsi
# -------------------------
ws = wb["IE40_rata2_prov"]

# Kolom B-F = 5 model
for row in ws.iter_rows(
    min_row=2,
    min_col=2,
    max_col=6
):
    for cell in row:
        cell.number_format = "0.00%"

wb.save(file_evaluasi)

print("Tersimpan:", file_evaluasi)
```
